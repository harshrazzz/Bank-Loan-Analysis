MTD- total Fund amount

```python
latest_issue_date = df['issue_date'].max()
latest_month = latest_issue_date.month
latest_year = latest_issue_date.year
mtd = df[(df['issue_date'].dt.month == latest_month) & (df['issue_date'].dt.year == latest_year)]
total_fund_mtd = mtd['loan_amount'].sum()
total_fund_mtd_millions = total_fund_mtd / 1e6
print("Total fund requested in the latest month in millions:${:,.2f} M".format(total_fund_mtd_millions))
```

Total Amount Received


```python
latest_issue_date = df['issue_date'].max()
latest_month = latest_issue_date.month
latest_year = latest_issue_date.year
mtd = df[(df['issue_date'].dt.month == latest_month) & (df['issue_date'].dt.year == latest_year)]
total_fund_mtd = mtd['total_payment'].sum()
total_fund_mtd_millions = total_fund_mtd / 1e6
print("Total payment received in the latest month in millions:${:,.2f} M".format(total_fund_mtd_millions))
```

```python
average_interest_rate = df['int_rate'].mean()*100
print("Average interest rate: {:.2f}%".format(average_interest_rate))   
```

```python
average_dti= df['dti'].mean()*100
print("Average DTI: {:.2f}%".format(average_dti))
```

```python
good_loan = df[df['loan_status'].isin(["Fully Paid","Current"])]

good_loan_count = good_loan['id'].count()
good_loan_percentage = (good_loan_count / LA) * 100
good_loan_amount = good_loan['loan_amount'].sum() /1e6
good_loan_received = good_loan['total_payment'].sum() /1e6
print(f"Number of good loans: {good_loan_count} ({good_loan_percentage:.2f}%)")
print(f"Total amount of good loans in millions: ${good_loan_amount:.2f} M")
print(f"Total payment received from good loans in millions: ${good_loan_received:.2f} M")
```

```python
bad_loan = df[df['loan_status'].isin(["Charged Off"])]

bad_loan_count = bad_loan['id'].count()
bad_loan_percentage = (bad_loan_count / LA) * 100
bad_loan_amount = bad_loan['loan_amount'].sum() /1e6
bad_loan_received = bad_loan['total_payment'].sum() /1e6
print(f"Number of bad loans: {bad_loan_count} ({bad_loan_percentage:.2f}%)")
print(f"Total amount of bad loans in millions: ${bad_loan_amount:.2f} M")
print(f"Total payment received from bad loans in millions: ${bad_loan_received:.2f} M")
```

Monthly Trend By issue date

```python
monthly_funded= ((df.sort_values('issue_date')).assign(month_name = lambda x: x['issue_date'].dt.strftime('%b %Y'))
                 .groupby('month_name',sort=False)['loan_amount'].sum().div(1e6).reset_index(name='total_fund_millions'))
plt.figure(figsize=(10,5))
plt.fill_between(monthly_funded['month_name'], monthly_funded['total_fund_millions'], color='skyblue', alpha=0.4)
plt.plot(monthly_funded['month_name'], monthly_funded['total_fund_millions'], marker='o', color='blue')
for i,row in monthly_funded.iterrows():
    plt.text(i, row['total_fund_millions']+ 0.1, f"{row['total_fund_millions']:.2f}", ha='center', va='bottom')  

#plt.bar(monthly_funded['month_name'], monthly_funded['total_fund_millions'], color='skyblue')
plt.xlabel('Month')
plt.ylabel('Total Fund Requested (Millions)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.xticks(rotation=45)
plt.show()
```

Total Amount Received

```python
monthly_received= ((df.sort_values('issue_date')).assign(month_name = lambda x: x['issue_date'].dt.strftime('%b %Y'))
                 .groupby('month_name',sort=False)['total_payment'].sum().div(1e6).reset_index(name='total_fund_millions'))
plt.figure(figsize=(10,5))
plt.fill_between(monthly_received['month_name'], monthly_received['total_fund_millions'], color='lightgreen', alpha=0.4)
plt.plot(monthly_received['month_name'], monthly_received['total_fund_millions'], marker='o', color='green')
for i,row in monthly_received.iterrows():
    plt.text(i, row['total_fund_millions']+ 0.1, f"{row['total_fund_millions']:.2f}", ha='center', va='bottom')  

#plt.bar(monthly_received['month_name'], monthly_received['total_fund_millions'], color='skyblue')
plt.xlabel('Month')
plt.ylabel('Total Fund Received (Millions)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.xticks(rotation=45)
plt.show()
```

```python
monthly_applications= ((df.sort_values('issue_date')).assign(month_name = lambda x: x['issue_date'].dt.strftime('%b %Y'))
                 .groupby('month_name',sort=False)['id'].count().reset_index(name='total_Loan_applications'))
plt.figure(figsize=(10,5))
plt.fill_between(monthly_applications['month_name'], monthly_applications['total_Loan_applications'], color='orange', alpha=0.4)
plt.plot(monthly_applications['month_name'], monthly_applications['total_Loan_applications'], marker='o', color='darkorange')
for i,row in monthly_applications.iterrows():
    plt.text(i, row['total_Loan_applications']+ 0.5, f"{row['total_Loan_applications']:.2f}", ha='center', va='bottom')  

#plt.bar(monthly_received['month_name'], monthly_received['total_fund_millions'], color='skyblue')
plt.xlabel('Month')
plt.ylabel('Total Loan Applications')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.xticks(rotation=45)
plt.show()
```

```python
state_wise_fund = df.groupby('address_state')['loan_amount'].sum().div(1000).sort_values(ascending=True)
plt.figure(figsize=(10,8))
barh = plt.barh(state_wise_fund.index, state_wise_fund.values, color='teal', alpha=0.6)
plt.xlabel('Total Fund Requested (Thousands)')
plt.ylabel('State')
plt.title('State-wise Fund Requested')
plt.tight_layout()
plt.show()
```

```python
state_wise_received = df.groupby('address_state')['loan_amount'].sum().div(1000).sort_values(ascending=True)
plt.figure(figsize=(10,8))
barh = plt.barh(state_wise_received.index, state_wise_received.values, color='purple', alpha=0.6)
plt.xlabel('Total Fund Received (Thousands)')
plt.ylabel('State')
plt.title('State-wise Fund Received')
plt.tight_layout()
plt.show()
```

Loan Term Analysis by Total Funded Amount


```python
term_wise_fund = df.groupby('term')['loan_amount'].sum().div(1e6).sort_values(ascending=True)
plt.figure(figsize=(5,5))
plt.pie(term_wise_fund.values, labels=term_wise_fund.index, 
        autopct=lambda p: f"{p:.1f}%\n${p*sum(term_wise_fund)/100:.1f}M", startangle=100,colors=['red','blue'])
plt.gca().add_artist(plt.Circle((0, 0), 0.70, color='white'))  # Add a white circle in the center to create a donut chart
plt.title('Fund Distribution by Loan Term') 
```

```python
emp_wise_fund = df.groupby('emp_length')['loan_amount'].sum().div(1000).sort_values(ascending=True)
plt.figure(figsize=(10,6))
barh = plt.barh(emp_wise_fund.index, emp_wise_fund.values, color='navy')
for bar in barh:
    width = bar.get_width()
    plt.text(width + 0.1, bar.get_y() + bar.get_height()/2, f"${width:.0f}K", va='center')
plt.xlabel('Total Fund Requested (Thousands)')  
plt.ylabel('Employment Length')
plt.tight_layout() 
plt.title('Fund Requested by Employment Length')    
```

```python
purpose_wise_fund = (df.groupby('purpose')['loan_amount'].sum().sort_values()/1e6)
plt.figure(figsize=(10,6))  
bars = plt.barh(purpose_wise_fund.index, purpose_wise_fund.values, color='coral')
for bar in bars:
    weight = bar.get_width()
    plt.text(weight + 0.1, bar.get_y() + bar.get_height()/2, f"${weight:.2f}M", va='center',fontsize=10)    
plt.xlabel('Total Fund Requested (Millions)')  
plt.ylabel('Loan Purpose')
plt.title('Fund Requested by Loan Purpose')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
home_ownership_fund = df.groupby('home_ownership')['loan_amount'].sum().reset_index()
home_ownership_fund['loan_amount'] = home_ownership_fund['loan_amount'] / 1e6
fig = px.treemap(home_ownership_fund, path=['home_ownership'], values='loan_amount', title='Fund Requested by Home Ownership', color ='loan_amount', color_continuous_scale='Blues')
fig.show()
```
